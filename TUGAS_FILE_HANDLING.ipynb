```python
import requests
import json

# Koordinat Bandar Lampung
latitude = -5.45
longitude = 105.27

# API Open Meteo (cuaca 7 hari)
url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&daily=temperature_2m_max,temperature_2m_min&timezone=Asia%2FBangkok"

# Ambil data dari API
response = requests.get(url)
data = response.json()

# Ambil data harian
dates = data["daily"]["time"]
temp_max = data["daily"]["temperature_2m_max"]
temp_min = data["daily"]["temperature_2m_min"]

# Simpan ke file (append)
with open("cuaca.txt", "a") as file:
    for i in range(len(dates)):  # LOOP
        tanggal = dates[i]
        max_t = temp_max[i]
        min_t = temp_min[i]

        # KONDISI (misal: simpan hanya suhu max > 30)
        if max_t > 30:
            hasil = f"{tanggal} | Max: {max_t}°C | Min: {min_t}°C\n"
            file.write(hasil)

print("Data berhasil disimpan!")
```
